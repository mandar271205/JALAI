# Copernicus DEM & Terrain Features Visualization

This notebook visualizes the raw DEM and computed terrain features (slope, flow accumulation, etc.) for the Mumbai pilot region.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import rasterio

ROOT = Path("../").resolve()
STATIC_DIR = ROOT / "data" / "processed" / "static"
```

```python
def plot_raster(path, title, cmap="viridis", vmin=None, vmax=None):
    if not path.exists():
        print(f"File not found: {path}")
        return
    with rasterio.open(path) as src:
        data = src.read(1)
        data = np.ma.masked_invalid(data)
        
        plt.figure(figsize=(8, 8))
        plt.imshow(data, cmap=cmap, vmin=vmin, vmax=vmax)
        plt.colorbar(label=title)
        plt.title(title)
        plt.show()
```

```python
plot_raster(STATIC_DIR / "elevation.tif", "Elevation (m)", cmap="terrain")
```

```python
plot_raster(STATIC_DIR / "slope.tif", "Slope (degrees)", cmap="magma")
```

```python
# Use logarithmic scale for flow accumulation
path = STATIC_DIR / "flow_accumulation.tif"
if path.exists():
    with rasterio.open(path) as src:
        data = src.read(1)
        data = np.log1p(np.ma.masked_invalid(data))
        plt.figure(figsize=(8, 8))
        plt.imshow(data, cmap="Blues")
        plt.colorbar(label="Log Flow Accumulation")
        plt.title("Flow Accumulation")
        plt.show()
```

```python
plot_raster(STATIC_DIR / "low_lying_index.tif", "Low Lying Index", cmap="Reds")
```
